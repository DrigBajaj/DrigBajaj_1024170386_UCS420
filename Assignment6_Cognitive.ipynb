**Q1 Vectorized Operations and Boolean Masking in NumPy**




```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Sensor Readings: ",temperature)
```

**a. Add 2°C to every reading using vectorization**

```python
corrected_temp = temperature + 2
print("Corrected temperatures (+2°C):", corrected_temp)
```

**b. Convert all temperatures from Celsius to Fahrenheit**

```python
temp_fahrenheit = (9 / 5) * corrected_temp + 32
print("Temperatures in Fahrenheit:", temp_fahrenheit)
```

**c. Identify all readings greater than 32°C using Boolean indexing**

```python
above_32 = corrected_temp[corrected_temp > 32]
print("Readings greater than 32°C:", above_32)
```

**d. Count how many readings exceed 32°C**

```python
count_above_32 = np.sum(corrected_temp > 32)
print("Number of readings exceeding 32°C:", count_above_32)
```

**e. Performance Analysis: Vectorization & Boolean Indexing vs. Iterative Loops**

**Compiled C-Level Execution:** Python's native for loops execute in an interpreted environment, causing repeated dynamic type-checking, reference counting, and interpreter overhead on every iteration. Vectorized operations and Boolean masking delegate execution directly to pre-compiled, optimized C/Fortran routines.

**Cache Locality and Contiguous Memory:** Unlike standard Python lists that store pointers to scattered objects, NumPy arrays reside in contiguous memory blocks with homogeneous data types (dtype). This maximizes CPU cache utilization and eliminates pointer dereferencing.

**SIMD Hardware Acceleration:** Vectorization allows modern processors to use Single Instruction, Multiple Data (SIMD) instruction sets. A single CPU instruction operates simultaneously on multiple data points per cycle rather than evaluating values sequentially.

**Optimized Filtering (Avoids Reallocation):** Boolean indexing evaluates conditions in a single pass at the hardware level and extracts matching entries directly. This avoids the memory reallocation and runtime cost of repeatedly calling list.append() inside an explicit loop.

**Q2 Multi-Dimensional Array Aggregations and Reductions in NumPy**

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

**a. Calculate the total steps recorded**

```python
total_steps = np.sum(steps)
print("Total steps recorded:", total_steps)
```

**b. Calculate the mean number of steps**

```python
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
```

**c. Find the maximum and minimum values**

```python
max_steps = np.max(steps)
min_steps = np.min(steps)

print("Maximum steps:", max_steps)
print("Minimum steps:", min_steps)
```

**d. Calculate the total steps for each day using axis=0**

```python
daily_totals = np.sum(steps, axis=0)
print("Total steps per day (Day 1, Day 2, Day 3):", daily_totals)
```

**e. Calculate the total steps for each user using axis=1**

```python
user_totals = np.sum(steps, axis=1)
print("Total steps per user (User 1 to 4):", user_totals)
```

**f. Find the user/day position containing the maximum number of steps using argmax()**

```python
# Flattened index
flat_idx = np.argmax(steps)
# Convert to 2D matrix coordinates (row=user, col=day)
user_idx, day_idx = np.unravel_index(flat_idx, steps.shape)
print("Flat index:", flat_idx)
print(f"2D Position (Index): User row {user_idx}, Day column {day_idx}")
print(f"Actual Location: User {user_idx + 1}, Day {day_idx + 1} with {steps[user_idx, day_idx]} steps")
```

**Q3 NumPy Array Slicing, Copy vs. View, and Shape Manipulation**

```python
# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a) Original array:", original)

# b. Create slice from index 1 to 4
subset = original[1:5]
print("b) Subset (index 1 to 4):", subset)

# c. Modify first element of subset to 999
subset[0] = 999
print("c) Subset after modifying first element:", subset)
print("   Original array (affected):", original)

# d. Create slice with copy and modify first element to 500
copied_slice = original[1:5].copy()
copied_slice[0] = 500
print("d) Copied slice after modification:", copied_slice)
print("   Original array (unaffected):", original)

# e. Create array 1 to 12 and reshape into 3x4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("e) 3x4 Matrix:\n", matrix)

# f. Extract rows, column, and submatrix
print("f) First row:", matrix[0, :])
print("   Last row:", matrix[-1, :])
print("   Second column:", matrix[:, 1])
print("   Elements from rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

# g. Flatten using flatten and ravel
flat_arr = matrix.flatten()
ravel_arr = matrix.ravel()
print("g) Flattened array using flatten():", flat_arr)
print("   Flattened array using ravel():", ravel_arr)

# h. Modify element in ravel array
ravel_arr[0] = 777
print("h) ravel_arr after modifying index 0:", ravel_arr)
print("   Original matrix after modifying ravel array (affected):\n", matrix)

# i. Modify element in flatten array
flat_arr[1] = 888
print("i) flat_arr after modifying index 1:", flat_arr)
print("   Original matrix after modifying flatten array (unaffected):\n", matrix)

# j. Display matrix attributes
print("j) Shape:", matrix.shape)
print("   Number of dimensions (ndim):", matrix.ndim)
print("   Total elements (size):", matrix.size)
print("   Data type (dtype):", matrix.dtype)
```

**Q4 Matrix Operations and Ordinary Least Squares (OLS) Regression in NumPy**

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

**a. Display the shape and dimensions of X**

```python
print("Shape of X:", X.shape)
print("Dimensions (ndim) of X:", X.ndim)
```

**b. Calculate: X.T and explain what the transpose represents**

```python
XT = X.T
print("Transpose of X (X.T):\n", XT)
```

**Explanation:**
The transpose operation swaps rows and columns. In the original matrix X, rows represent individual users and columns represent features. In X.T, each row represents one feature (sleep, activity, stress) across all users, and each column represents a single user.

**c. Calculate the matrix product: X.T @ X**

```python
XTX = X.T @ X
print("Matrix product (X.T @ X):\n", XTX)
```

**d. Calculate the matrix inverse using np.linalg.inv() where appropriate**

```python
XTX_inv = np.linalg.inv(XTX)
print("Inverse of (X.T @ X):\n", XTX_inv)
```

**e. Implement the Ordinary Least Squares equation: $\beta = (X^T X)^{-1} X^T y$ using NumPy**

```python
beta = XTX_inv @ X.T @ y
print("OLS Coefficients (beta):\n", beta)
```

**f. Explain what the resulting coefficients represent in the context of the three user features**

**$\beta_0$ (Sleep Hours $\approx -0.16$):** A negative relationship. For each additional hour of sleep, the required assistance score decreases by approximately 0.16 units, holding other factors constant.

**$\beta_1$ (Activity Level $\approx +0.14$):** A slight positive relationship. For each 1-unit increase in daily activity, the assistance score increases by approximately 0.14 units.

**$\beta_2$ (Stress Level $\approx +10.06$):** The primary contributing factor. For each 1-unit increase on the stress scale, the assistance score increases sharply by roughly 10.06 units.

**g. Predict score for new user**

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", round(predicted_score, 2))
```
