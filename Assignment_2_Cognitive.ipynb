**Q1**


**Create List**

```python
rno="1024170386"
L=[]
for digit in rno:
  L.append(int(digit)*10)
```

**(i) Print L**

```python
print(L)
```

**(ii) Add elements using append() and insert()**

```python
# Using append()
L.append(55)
print("After append(55):", L)
# append() adds the element at the end of the list.

# Using insert()
L.insert(3, 99)
print("After insert(3, 99):", L)
# insert(index, value) adds the element at the specified position.
```

**(iii) Remove elements using remove() and pop()**

```python
# Using remove()
L.remove(55)
print("After remove(55):", L)
# remove(value) removes the first occurrence of the given value.

# Using pop()
removed = L.pop(2)
print("After pop(2):", L)
print("Removed element:", removed)
# pop(index) removes and returns the element at the specified index.
```

**(iv) Sort the list**

```python
# Ascending order
L.sort()
print("Ascending Order:", L)

# Descending order
L.sort(reverse=True)
print("Descending Order:", L)
```

**(v) Slicing**

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

**(vi) List comprehension**

```python
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]
print("Average =", average)
print("Elements greater than average:", greater_than_average)
```

**Q3**

**(i) Import random and generate the list**

```python
import random
# Set seed using your roll number
random.seed(1024170386)
# Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for i in range(100)]
print(numbers)
```

**(ii) Count and print odd numbers**

```python
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd Numbers:")
print(odd_numbers)
print("Count of Odd Numbers:", len(odd_numbers))
```

**(iii) Count and print even numbers**

```python
even_numbers = [num for num in numbers if num % 2 == 0]
print("Even Numbers:")
print(even_numbers)
print("Count of Even Numbers:", len(even_numbers))
```

**(iv) Count and print prime numbers**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
# List comprehension to build the list of prime numbers
prime_numbers = [num for num in numbers if is_prime(num)]
print("Prime Numbers:")
print(prime_numbers)
print("Count of Prime Numbers:", len(prime_numbers))
```

**(v) Most frequent number**

```python
max_count = 0
most_frequent = None
for num in numbers:
    count = numbers.count(num)
    if count > max_count:
        max_count = count
        most_frequent = num
print("Most Frequent Number:", most_frequent)
print("Number of Occurrences:", max_count)
```

**Q4**

**Create sets A and B**

```python
# First 8 digits of the roll number
digits = [1, 0, 2, 4, 1, 7, 0, 3]
# Create sets
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)
```

**(vi) Union**

```python
union_set = A.union(B)
print("Union of A and B:", union_set)
```

**(vii) Intersection**

```python
intersection_set = A.intersection(B)
print("Intersection of A and B:", intersection_set)
```

**(viii) Difference**

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
# difference() returns elements present in one set but not the other,
# whereas symmetric_difference() returns elements that are in either set but not in both.
```

**(ix) Symmetric Difference**

```python
sym_diff = A.symmetric_difference(B)
print("Symmetric Difference:", sym_diff)
```

**(x) Subset and Superset**

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**(xi) Remove user-entered value using discard()**

```python
x = int(input("Enter a value to remove from Set A: "))
A.discard(x)
print("Updated Set A:", A)
# discard() does not raise an error if the value is not present,
# whereas remove() raises a KeyError.
```
