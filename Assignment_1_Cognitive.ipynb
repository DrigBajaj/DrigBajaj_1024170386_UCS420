**Assingment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.**

```python
n=int(input("Enter a number:"))
sum=0
for i in range(1,n+1):
  if(i%7==0) and (i%9==0):
    sum=sum+i
print("The sum of numbers from 1 to ",n," divisible by 7 and 9 is: ",sum)
```

**Assingment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.**

```python
n=int(input("Enter a number: "))
sum=0
for i in range(2,n+1):
    for j in range(2,int(i**0.5)+1):
        if i%j==0:
            break
    else:
        sum+=i
print("The sum of prime numbers from 1 to", n, "is", sum)
```

**Assingment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.**

```python
def AddOdd(num):
  sum=0
  for i in range(1,num+1):
    if(i%2!=0):
      sum+=i
  return sum
n=int(input("Enter a number: "))
print("The sum of odd numbers from 1 to ",n," is: ",AddOdd(n))
```

**Assingment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.**

```python
def AddPrime(num):
  sum=0
  for i in range(2,num+1):
      for j in range(2,int(i**0.5)+1):
          if i%j==0:
              break
      else:
          sum+=i
  return sum
n=int(input("Enter a number: "))
print("The sum of prime numbers from 1 to", n, "is", AddPrime(n));
```
