# **Q1**

```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Cricket is my favorite sport because it is exciting and entertaining.
I enjoy watching cricket matches with my friends and family.
The players show great teamwork, patience, and determination.
Batting and bowling make every match interesting.
I especially enjoy watching close matches that go down to the final over.
Cricket brings people together and teaches us the value of teamwork."""
```

**Q1(**i**) Convert Text to Lowercase and Remove Punctuation Using re**

```python
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Cleaned Text:\n", clean_text)
```

**Q1(ii). Tokenize the Text into Words and Sentences**

```python
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)
print("Word Tokens:\n", words)
print("\nSentence Tokens:\n", sentences)
```

**Q1(iii) Split Using split() and word_tokenize() and Compare**

```python
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("Using split():\n", split_words)
print("\nUsing word_tokenize():\n", nltk_words)

print("\nNumber of tokens using split():", len(split_words))
print("Number of tokens using word_tokenize():", len(nltk_words))
```

**Comparison:** split() separates words based on whitespace, while word_tokenize() uses NLTK's tokenization rules. Since punctuation has already been removed, both methods produce similar results for this paragraph.

**Q1(iv) Remove Stopwords Using NLTK**

```python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in nltk_words
    if word not in stop_words
]
print("Words after removing stopwords:\n", filtered_words)
```

**Q1(v) Display Word Frequency Distribution Excluding Stopwords**

```python
word_freq = Counter(filtered_words)

print("Word Frequency Distribution:")
for word, frequency in word_freq.most_common():
    print(word, ":", frequency)
```

# **Q2**

**Q2(i) Extract All Words Containing Only Alphabets Using re.findall()**

```python
words = re.findall(r'\b[a-zA-Z]+\b', clean_text)

print("Words containing only alphabets:\n", words)
```

**Q2(ii) Remove Stopwords Using NLTK**

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("Words after removing stopwords:\n", filtered_words)
```

**Q2(iii) Perform Stemming Using PorterStemmer**

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]

print("Stemmed Words:\n", stemmed_words)
```

**Q2(iv) Perform Lemmatization Using WordNetLemmatizer**

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Lemmatized Words:\n", lemmatized_words)
```

**Q2(v) Compare Stemmed and Lemmatized Outputs**

```python
print("Original Words:", filtered_words)
print("Stemmed Words:", stemmed_words)
print("Lemmatized Words:", lemmatized_words)
```

**Explanation:**
- **Stemming:** Removes word endings to obtain a root-like form. The result may not be a meaningful English word. For example, matches becomes match.
- **Lemmatization:** Converts words to their meaningful dictionary form. For example, players becomes player.

I would prefer stemming when speed is important, such as in basic text searches. I would prefer lemmatization when meaningful words and better readability are important, such as in text analysis and natural language processing.

# **Q3**

```python
texts = [
    "The government announced new policies to improve education.",
    "Heavy rainfall caused flooding in several cities yesterday.",
    "Scientists developed a new battery for electric vehicles."
]

print("Texts:")
for i, text in enumerate(texts, 1):
    print(f"Text {i}: {text}")
```

**Q3(i) Generate Bag of Words Using CountVectorizer**

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
bow_matrix = cv.fit_transform(texts)

print("Vocabulary:")
print(cv.get_feature_names_out())

print("\nBag of Words Representation:")
print(bow_matrix.toarray())
```

**Q3(ii) Compute TF-IDF Scores Using TfidfVectorizer**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(texts)

print("TF-IDF Feature Names:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Scores:")
print(tfidf_matrix.toarray().round(3))
```

**Q3(iii) Print and Interpret the Top 3 Keywords from Each Text Using TF-IDF**

```python
feature_names = tfidf.get_feature_names_out()
scores = tfidf_matrix.toarray()

for i, row in enumerate(scores):
    top_indices = row.argsort()[-3:][::-1]

    print(f"\nText {i + 1}:")
    for index in top_indices:
        print(feature_names[index], ":", round(row[index], 3))
```

**Interpretation**
- **Text 1 — Education:** The top keywords identify the government's policies and education-related announcement.
- **Text 2 — Weather:** The top keywords highlight rainfall, flooding, and affected cities.
- **Text 3 — Technology:** The top keywords emphasize scientists, batteries, and electric vehicles.

**Conclusion:** CountVectorizer represents text using word counts, while TfidfVectorizer assigns scores based on word importance across the collection of texts. Since the three texts cover different topics, TF-IDF helps identify words that distinguish each text.

# **Q4**

```python
text1 = """Artificial Intelligence enables machines to perform tasks.
It can learn from data and recognize patterns.
AI is used in chatbots and recommendation systems.
It helps automate tasks and solve complex problems."""

text2 = """Blockchain stores data in a distributed digital ledger.
It links records together using cryptographic techniques.
Blockchain is used in cryptocurrency and secure transactions.
It improves transparency and makes records difficult to alter."""

print("Text 1 (Artificial Intelligence):\n", text1)
print("\nText 2 (Blockchain):\n", text2)
```

**Q4(i) Preprocess and Tokenize Both Texts**

```python
clean_text1 = re.sub(r'[^\w\s]', '', text1.lower())
clean_text2 = re.sub(r'[^\w\s]', '', text2.lower())

tokens1 = word_tokenize(clean_text1)
tokens2 = word_tokenize(clean_text2)

print("AI Tokens:\n", tokens1)
print("\nBlockchain Tokens:\n", tokens2)
```

**Q4(ii)(a) Calculate Jaccard Similarity Using Sets**

```python
set1 = set(tokens1)
set2 = set(tokens2)
jaccard_similarity = len(set1 & set2) / len(set1 | set2)

print("Jaccard Similarity:", round(jaccard_similarity, 3))
```

**Q4(ii)(b) Calculate Cosine Similarity Using TfidfVectorizer**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

texts = [clean_text1, clean_text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("Cosine Similarity:", round(cosine_sim[0][0], 3))
```

**Q4(ii)(c) Analyze Which Similarity Metric Gives Better Insights**

**Analysis:**
- **Jaccard similarity** is useful for comparing the overlap between unique words in the two texts.
- **Cosine similarity** considers TF-IDF weights, giving more importance to words that are distinctive within the collection.
- In this case, **cosine similarity provides slightly richer information** because it considers word importance, while Jaccard treats every unique word equally.

# **Q5**

```python
reviews = [
    "This smartphone has an excellent camera and a beautiful display. I love its performance.",
    "The battery life is terrible and the phone gets hot quickly. I am very disappointed.",
    "The phone is average and has basic features. It works as expected.",
    "The design is amazing and the battery lasts all day. I highly recommend this phone.",
    "The phone is slow and the customer service was very poor."
]

for i, review in enumerate(reviews, 1):
    print(f"Review {i}: {review}")
```

**Q5(i) Find Polarity and Subjectivity Using TextBlob**

```python
from textblob import TextBlob
import pandas as pd

results = []

for review in reviews:
    analysis = TextBlob(review).sentiment

    results.append({
        "Review": review,
        "Polarity": round(analysis.polarity, 3),
        "Subjectivity": round(analysis.subjectivity, 3)
    })

df = pd.DataFrame(results)
print(df.to_string(index=False))
```

**Q5(ii) Classify Reviews into Positive, Negative, or Neutral**

```python
def classify_sentiment(polarity):
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Polarity"].apply(classify_sentiment)

print(df[["Review", "Polarity", "Subjectivity", "Sentiment"]].to_string(index=False))
```

**Q5(iii) Create a Word Cloud for All Positive Reviews**

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

positive_reviews = df[df["Sentiment"] == "Positive"]["Review"]

positive_text = " ".join(positive_reviews)

if positive_text.strip():
    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews found.")
```

```python
text = """boAt headphones are popular for their stylish design and affordable prices.
They provide clear sound quality for music and videos.
Many models offer powerful bass for an enjoyable listening experience.
Wireless headphones connect easily to smartphones through Bluetooth.
Their long battery life makes them suitable for everyday use.
Some models include noise cancellation to reduce background sounds.
The comfortable ear cushions allow users to listen for longer periods.
boAt headphones are available in different colors and designs.
They are useful for studying, travelling, gaming, and entertainment.
Overall, boAt headphones offer a convenient audio experience for everyday users."""
```

**Q6(i) Tokenize Text Using Tokenizer() from Keras**

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Word Index:")
print(tokenizer.word_index)

print("\nTotal Words:", total_words)
```

**Q6(ii) Create Input Sequences**

```python
input_sequences = []

for line in text.split('.'):
    sequence = tokenizer.texts_to_sequences([line.strip()])[0]

    for i in range(1, len(sequence)):
        input_sequences.append(sequence[:i + 1])

max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Input Shape:", X.shape)
print("Output Shape:", y.shape)
```

**Q6(ii) Build and Train a Simple LSTM Model**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(total_words, 32),
    LSTM(64),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(X, y, epochs=100, verbose=1)
```
