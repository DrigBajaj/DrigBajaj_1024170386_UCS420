**Q1 NumPy 1D Array – Basic Operations**

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print("Original Array:", arr)

# a) Add 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply all elements by 3
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

**Q2 Basic NumPy Array Operations**

**a) Reverse the NumPy array**

```python
arr = np.array([1, 2, 3, 6, 4, 5])
# Reverse the array
reverse_arr = np.flip(arr)
print("Original Array:", arr)
print("Reversed Array:", reverse_arr)
```

**b) Find the most frequent value and its indices**

**i. Array x**

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# Find unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

# Find the most frequent value
most_frequent = values[np.argmax(counts)]

# Find indices of the most frequent value
indices = np.where(x == most_frequent)[0]

print("Array x:", x)
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

**ii. Array y**



```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

# Find the highest frequency
max_count = np.max(counts)

# Find all values having the highest frequency
most_frequent = values[counts == max_count]

print("Array y:", y)
print("Most frequent values:", most_frequent)

# Find indices of each most frequent value
for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

**Q3 Access Elements from a 2-D NumPy Array**

```python
arr = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])

# a) Access 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

**Q4 NumPy Array using linspace()**

```python
# Create 1-D array with 25 evenly spaced numbers from 10 to 100
Drig_Bajaj = np.linspace(10, 100, 25)
print("Array:")
print(Drig_Bajaj)

# Dimensions of the array
print("Dimensions:", Drig_Bajaj.ndim)

# Shape of the array
print("Shape:", Drig_Bajaj.shape)

# Total number of elements
print("Total elements:", Drig_Bajaj.size)

# Data type of each element
print("Data type:", Drig_Bajaj.dtype)

# Total number of bytes consumed
print("Total bytes:", Drig_Bajaj.nbytes)

# Transpose using reshape()
transpose = Drig_Bajaj.reshape(25,1)
print("Transpose using reshape():")
print(transpose)

# Transpose using T attribute
print("Transpose using T:")
print(Drig_Bajaj.reshape(1, 25).T)
```

**Answer:** Yes, we can use the T attribute. However, for a 1-D array, Drig_Bajaj.T alone does not change its shape. We need to reshape it first.

**Q5 2-D NumPy Array Operations**

```python
# Create a 2-D array with 3 rows and 4 columns
ucs420_Drig_Bajaj = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])
print("Original Array:")
print(ucs420_Drig_Bajaj)

# Mean
print("Mean:", np.mean(ucs420_Drig_Bajaj))

# Median
print("Median:", np.median(ucs420_Drig_Bajaj))

# Maximum
print("Maximum:", np.max(ucs420_Drig_Bajaj))

# Minimum
print("Minimum:", np.min(ucs420_Drig_Bajaj))

# Unique elements
print("Unique Elements:", np.unique(ucs420_Drig_Bajaj))

# Reshape into 4 rows and 3 columns
reshaped_ucs420_Drig_Bajaj = ucs420_Drig_Bajaj.reshape(4, 3)
print("\nReshaped Array:")
print(reshaped_ucs420_Drig_Bajaj)

# Resize into 2 rows and 3 columns
resized_ucs420_Drig_Bajaj = np.resize(ucs420_Drig_Bajaj, (2, 3))
print("\nResized Array:")
print(resized_ucs420_Drig_Bajaj)
```
